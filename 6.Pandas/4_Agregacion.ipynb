## Varias métricas por grupo

```python
resumen = (
    ventas
    .groupby("vendedor")["monto"]
    .agg(["sum", "mean", "max", "min", "count"])
)

print(resumen)
```

⭐ Reto: ¿quién tiene el mejor desempeño?

- Calcular el total vendido por producto

- Obtener por producto:

-número de ventas  
-venta total  
-promedio  
-venta máxima

agg() con nombres personalizados

```python
resumen = (
    ventas
    .groupby("vendedor")
    .agg(
        ventas_totales=("monto", "sum"),
        venta_promedio=("monto", "mean"),
        venta_maxima=("monto", "max"),
        numero_ventas=("monto", "count")
    )
)

print(resumen)
```
