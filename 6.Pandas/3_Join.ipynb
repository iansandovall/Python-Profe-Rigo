```python
import pandas as pd

# Crear el primer DataFrame
import pandas as pd

ventas = pd.DataFrame({
    "Empleado":           ["Ana", "Luis", "María", "Pedro", "Sofía"],
    "ventas":             [120, 250, 180, 90, 310],
    "devoluciones":       [5,   12,  0,   8,  3],
    "n_clientes_nuevos":  [3,   7,   4,   1,  9],
    "dias_trabajados":    [20,  18,  22,  15, 21],
    "region":             ["Norte", "Sur", "Norte", "Sur", "Norte"]
})

metas = pd.DataFrame({
    "Empleado":           ["Ana", "Luis", "María", "Pedro", "Sofía"],
    "meta_ventas":     [100, 300, 150, 120, 280],
    "meta_clientes":   [2,   6,   3,   1,   8],
    "bono_base":       [50,  100, 75,  60,  95],
    "meta_presencial": [18,  15,  18,  12,  20],
    "nivel":           ["junior", "senior", "medio", "junior", "senior"]
})

print("DataFrame de ventas:")
print(ventas)
print("\nDataFrame de metas:")
print(metas)
```

```python
# Usar la columna "Empleado" como índice en ambos DataFrames
ventas.set_index("Empleado", inplace=True)
metas.set_index("Empleado", inplace=True)
print("DataFrame de ventas:")
print(ventas)
print("\nDataFrame de metas:")
print(metas)
```

```python
#Unir los DataFrames usando un join
resultado = ventas.join(metas)
print("\nDataFrame unido: ")
display(resultado)
```

join() es especialmente cómodo cuando la información que relaciona las tablas ya está en los índices.

## Derivar columnas de cumplimiento

```python
resultado["cumplio_meta_ventas"] = resultado["ventas"] >= resultado["meta_ventas"]

resultado[["ventas", "meta_ventas", "cumplio_meta_ventas"]]
```

```python
##Hacerlo para meta clientes
resultado["cumplio_meta_clientes"] = resultado["n_clientes_nuevos"] >= resultado["meta_clientes"]
#margen de devoluciones
resultado["margen_devoluciones"] = resultado["devoluciones"] / resultado["ventas"]

resultado[["ventas", "meta_ventas", "cumplio_meta_ventas",
           "n_clientes_nuevos", "cumplio_meta_clientes", "devoluciones", "margen_devoluciones"]]
```

Crear el dataframe dos nuevamente, pero modificando el primer valor: ventas_lu, para forzar error

```python
ventas_1 = pd.DataFrame({
    "ventas_lu": [random.randint(200, 500) for _ in range(5)],
    "ventas_ma": [random.randint(200, 500) for _ in range(5)],
    "ventas_mi": [random.randint(200, 500) for _ in range(5)]


}, index=["Ana", "Luis", "María", "Pedro", "Sofía"])

ventas_2 = pd.DataFrame({
    "ventas_lu": [random.randint(200, 500) for _ in range(5)],
    "ventas_vi": [random.randint(200, 500) for _ in range(5)],
    "ventas_sa": [random.randint(200, 500) for _ in range(5)]
}, index=["Ana", "Luis", "Ángel", "Adrián", "Camila"])

print(ventas_1)
print(ventas_2)
```

```python
# Realizar la unión utilizando join
ventas_Semana = ventas_1.join(ventas_2, how='inner')  # Usando 'inner' para obtener la intersección de índices

ventas_Semana
```

Ambos DataFrames contienen una columna llamada ventas_lu.

Cuando haces join, Pandas no sabe cómo renombrar automáticamente esas columnas duplicadas.

```python
ventas_Semana = ventas_1.join(ventas_2,  lsuffix='_df1', rsuffix='_df2') 
ventas_Semana
```

```python
ventas_Semana= ventas_1.join(ventas_2, on = 'ventas_lu')
ventas_Semana
```

El método .join() está diseñado principalmente para unir por índices.

El parámetro on= se puede usar solo si el DataFrame a la izquierda (ventas_1) tiene esa columna (en este caso ventas_lu) y el DataFrame de la derecha (ventas_2) usa como índice la columna equivalente.

*Pero aquí hay dos problemas:

ventas_2 tiene como índice key, no ventas_lu.

on='ventas_mi' intenta alinear usando una columna que no corresponde al índice del otro DataFrame.

```python
ventas_reset = ventas_2.reset_index().set_index('ventas_lu')
print(ventas_reset)
ventas_Semana = ventas_1.join(ventas_reset, on='ventas_lu')
print(ventas_Semana)
```
