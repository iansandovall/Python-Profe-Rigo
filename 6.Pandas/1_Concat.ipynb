## Concatenar dataframes con columnas diferentes

```python
random.seed(42) #Semilla para reproducibilidad del experimento
sucursal_norte = pd.DataFrame({
    "Laptop" : [random.randint(1,20) for _ in range(3)],
    "Teclado" : [random.randint(1,20) for _ in range(3)],
    "Tablet" : [random.randint(1,20) for _ in range(3)],
    "Audifonos" : [random.randint(1,20) for _ in range(3)],
})
sucursal_sur = pd.DataFrame({
    "Laptop" : [random.randint(1,20) for _ in range(4)],
    "Mouse" : [random.randint(1,20) for _ in range(4)],
    "HDMI" : [random.randint(1,20) for _ in range(4)],
    "MicroSD" : [random.randint(1,20) for _ in range(4)],
})
display("Sucursal norte")
display(sucursal_norte)
display("Sucursal sur")
display(sucursal_sur)
```

```python
ventas = pd.concat([sucursal_norte, sucursal_sur], ignore_index= True)
ventas
#Une los datos a lo largo del índice y los datos que faltan se rellenan con NaN
```

```python
ventas = pd.concat([sucursal_norte, sucursal_sur], axis = 1)
ventas
```

```python
print(sucursal_norte)
print(sucursal_sur)
#Conservar las filas con índices coincidentes
ventas = pd.concat([sucursal_norte, sucursal_sur], axis = 1, join = 'inner')
print(ventas)
```

```python
sucursal_A = pd.DataFrame({
    "Producto" : ["Laptop", "Tablet", "Smartphone"],
    "Ventas_A" : [25, 85, 31]
})
sucursal_B = pd.DataFrame({
    "Producto" : ["Laptop", "Tablet", "Smartwatch"],
    "Ventas_B" : [42, 23, 40]
})
print(sucursal_A)
print(sucursal_B)
```

```python
ventas = pd.concat([sucursal_A.set_index('Producto'), sucursal_B.set_index('Producto')], axis = 1, join = 'outer')
ventas = ventas.fillna(0)
#ventas = ventas.astype(int)
print(ventas)
```

```python
sucursal_A = pd.DataFrame({
    "Producto" : ["Laptop", "Tablet", "Smartphone"],
    "Ventas" : [25, 85, 31]
})
sucursal_B = pd.DataFrame({
    "Producto" : ["Laptop", "Tablet", "Smartphone"],
    "Ventas" : [5, 12, 78]
})
sucursal_C = pd.DataFrame({
    "Producto" : ["Laptop", "Tablet", "Smartphone"],
    "Ventas" : [14, 50, 38]
})
print(sucursal_A)
print(sucursal_B)
print(sucursal_C)
```

```python
ventas = pd.concat([sucursal_A, sucursal_B, sucursal_C], keys= ["S_A", "S_B", "S_C"])
print(ventas)
```
