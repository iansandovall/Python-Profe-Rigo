Responde los siguientes puntos con código en Python:

1. Leer el archivo Calificaciones.txt y dividirlo en una lista de líneas. 

```python
#Aquí primero abrimos el archivo en modo lectura y leemos todas las líneas, luego cerramos el archivo y finalmente imprimimos las líneas leídas.
archivo = open("Calificaciones.txt", "r", encoding="utf-8")
lineas = archivo.readlines()
archivo.close()
print(lineas)
```

2. Procesar las líneas para obtener nombres y calificaciones en listas separadas.

```python
#En esta parte se crean dos listas vacías, una para los nombres y otra para las calificaciones. Y Luego, se cambian sobre cada línea del archivo, se divide la línea en partes (nombre y calificación), y se agregan a las listas correspondientes. Finalmente, se imprimen las listas de nombres y calificaciones.
nombres = []
calificaciones = []

for linea in lineas:
    partes = linea.strip().split()

    nombre = partes[0]
    calificacion = int(partes[1])

    nombres.append(nombre)
    calificaciones.append(calificacion)

print(nombres)
print(calificaciones)
```

3. Imprimir todas las calificaciones con su respectivo nombre. 

```python
#En este punto, se recorre la lista de calificaciones y se imprime el nombre del estudiante junto con su calificación correspondiente.
for i in range(len(calificaciones)):
    print(nombres[i], calificaciones[i])
```

4. Calcular el promedio de calificaciones.

```python
#Acá se calcula el promedio de las calificaciones sumando todas las calificaciones y dividiendo entre la cantidad de calificaciones. Luego se imprime el promedio.
promedio = sum(calificaciones) / len(calificaciones)
print("Promedio de calificaciones:", promedio)
```

5. Encontrar la calificación más alta y el estudiante que la obtuvo.

```python
#Aquí se encuentra la calificación más alta utilizando la función max() y se imprime. Luego, se encuentra el índice de la calificación más alta en la lista de calificaciones y se utiliza ese índice para imprimir el nombre del estudiante que obtuvo la calificación más alta.
max_calificacion = max(calificaciones)
print("Calificación más alta:", max_calificacion)

indice_max = calificaciones.index(max_calificacion)

print("Estudiante con la calificación más alta:", nombres[indice_max])
```

6. Mostrar la lista de estudiantes reprobados (<70).

```python
#Aquí se crea una lista vacía llamada reprobados y se recorre la lista de calificaciones. Si la calificación de un estudiante es menor a 70, se agrega su nombre a la lista de reprobados. Para despues imprimir la lista con los estudiantes reprobados.
reprobados = []
for i in range(len(calificaciones)):
    if calificaciones[i] < 70:
        reprobados.append(nombres[i])
print("Estudiantes reprobados:", reprobados)
```

*Parte 2: Análisis de Pedidos y Reseñas de una Tienda en Línea*

Descripción: Trabajas en el departamento de análisis de datos de una tienda en línea que vende productos electrónicos.
La empresa requiere conocer distintas métricas. Para esto, cuentas con dos bases de datos:

pedidos.csv → Contiene información de los pedidos realizados en la tienda.

reseñas.csv → Contiene las reseñas que los clientes han dejado sobre los productos comprados.

**Instrucciones**

Carga ambos archivos CSV en DataFrames.
Identificar la columna que relaciona ambas bases y realizar la unión apropiada.

```python
import pandas as pd

# Primero se cargan los archivos CSV y utilizamos lo de ISO-8859-1.
pedidos = pd.read_csv("pedidos.csv", encoding="ISO-8859-1")
reseñas = pd.read_csv("reseñas.csv", encoding="ISO-8859-1")

# Después se unen los DataFrames utilizando la columna de pedido id y de reseña id, lo que importa ahí es el ID por eso al final lo uso como "pedido_id" y con merge, especificando el tipo de unión que queremos (en este caso, outer join).
datos = pd.merge(pedidos, reseñas, on="pedido_id", how="outer")

# Y finalmente se imprime el DataFrame unido.
print(datos)
```

Responde las siguientes preguntas mediante el tipo de unión correspondiente y sentencias “print” :

¿Cuántos pedidos tienen reseñas? Imprimir cuántos son. 

```python
# Primero se unen únicamente los pedidos que tienen una reseña
pedidos_con_reseña = pd.merge(
    pedidos,
    reseñas,
    on="pedido_id",
    how="inner"
)

# Después se cuentan los pedidos que tienen reseña
cantidad_pedidos_con_reseña = len(pedidos_con_reseña)

# Y al final se imprime el resultado
print("Cantidad de pedidos con reseña:", cantidad_pedidos_con_reseña)
```

¿Cuántos pedidos no tienen una reseña asociada? Imprimir cuántos son.

```python
# Se unen solamente los pedidos que tienen una reseña con inner, y luego se resta la cantidad de pedidos con reseña al total de pedidos para obtener la cantidad de pedidos sin reseña
pedidos_con_reseña = pd.merge(
    pedidos,
    reseñas,
    on="pedido_id",
    how="inner"
)

# Después se restan los pedidos con reseña al total de pedidos con len para obtener la cantidad de pedidos sin reseña
cantidad_pedidos_sin_reseña = len(pedidos) - len(pedidos_con_reseña)

# Y al final se imprime el resultado
print("Cantidad de pedidos sin reseña:", cantidad_pedidos_sin_reseña)
```

¿Existen reseñas sin un pedido registrado en la base de datos?, si es así, imprimir cuáles son. 

```python
#Aquí se unen todos los pedidos con todas las reseñas utilizando un right join para conservar todas las reseñas, aunque alguna no tenga un pedido registrado, después se revisa cada pedido_id de las reseñas y se compara con los pedido_id de la lista de pedidos, si no existe, se agrega a la lista de reseñas sin pedido y finalmente se imprime
reseñas_unidas = pd.merge(
    pedidos,
    reseñas,
    on="pedido_id",
    how="right"
)

reseñas_sin_pedido = []

for pedido_id in reseñas_unidas["pedido_id"]:
    if pedido_id not in pedidos["pedido_id"].values:
        reseñas_sin_pedido.append(pedido_id)

print("Reseñas sin pedido registrado:", reseñas_sin_pedido)
```
