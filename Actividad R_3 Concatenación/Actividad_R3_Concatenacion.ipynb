## EJERCICIO 1 | Ventas por trimestre

Una empresa tiene las ventas de tres trimestres en DataFrames separados. Hay que unirlos con `pd.concat` y sacar las ventas totales de cada producto.

```python
import pandas as pd
```

```python
# Creamos un DataFrame por cada trimestre
ventas_q1 = pd.DataFrame({
    "order_id": [101, 102, 103],
    "product_id": [1, 2, 3],
    "product_name": ["Monitor", "Audífonos", "Smartwatch"],
    "units_sold": [6, 15, 8],
    "revenue": [27000, 13500, 20000],
    "quarter": ["Q1", "Q1", "Q1"]
})

ventas_q2 = pd.DataFrame({
    "order_id": [201, 202, 203],
    "product_id": [1, 2, 3],
    "product_name": ["Monitor", "Audífonos", "Smartwatch"],
    "units_sold": [4, 20, 11],
    "revenue": [18000, 18000, 27500],
    "quarter": ["Q2", "Q2", "Q2"]
})

ventas_q3 = pd.DataFrame({
    "order_id": [301, 302, 303],
    "product_id": [1, 2, 3],
    "product_name": ["Monitor", "Audífonos", "Smartwatch"],
    "units_sold": [9, 12, 5],
    "revenue": [40500, 10800, 12500],
    "quarter": ["Q3", "Q3", "Q3"]
})

display(ventas_q1)
display(ventas_q2)
display(ventas_q3)
```

```python
# Unimos los tres trimestres en un solo DataFrame
# ignore_index=True para que el índice quede corrido del 0 al 8
ventas_trimestres = pd.concat([ventas_q1, ventas_q2, ventas_q3], ignore_index=True)

display(ventas_trimestres)
```

```python
# Filtramos las ventas de cada producto por separado
monitor = ventas_trimestres[ventas_trimestres["product_name"] == "Monitor"]
audifonos = ventas_trimestres[ventas_trimestres["product_name"] == "Audífonos"]
smartwatch = ventas_trimestres[ventas_trimestres["product_name"] == "Smartwatch"]

display(monitor)
display(audifonos)
display(smartwatch)
```

```python
# Sumamos las unidades vendidas y los ingresos de cada producto
unidades_monitor = monitor["units_sold"].sum()
ingresos_monitor = monitor["revenue"].sum()

unidades_audifonos = audifonos["units_sold"].sum()
ingresos_audifonos = audifonos["revenue"].sum()

unidades_smartwatch = smartwatch["units_sold"].sum()
ingresos_smartwatch = smartwatch["revenue"].sum()

print("Monitor:", unidades_monitor, "unidades -", ingresos_monitor, "de ingresos")
print("Audífonos:", unidades_audifonos, "unidades -", ingresos_audifonos, "de ingresos")
print("Smartwatch:", unidades_smartwatch, "unidades -", ingresos_smartwatch, "de ingresos")
```

```python
# Armamos el DataFrame final con los totales de cada producto
resultado = pd.DataFrame({
    "product_name": ["Monitor", "Audífonos", "Smartwatch"],
    "total_units": [unidades_monitor, unidades_audifonos, unidades_smartwatch],
    "total_revenue": [ingresos_monitor, ingresos_audifonos, ingresos_smartwatch]
})

# Ordenamos de mayor a menor por unidades vendidas
resultado = resultado.sort_values("total_units", ascending=False)

display(resultado)
```

## EJERCICIO 2 | Ventas y clientes con merge

Una tienda en línea tiene sus ventas en `ventas.csv` y sus clientes en `clientes.csv`. Hay que unir las dos tablas con `merge()` usando los distintos tipos de join.

```python
# Cargamos los dos archivos csv
ventas = pd.read_csv("ventas.csv")

# clientes.csv marcaba error por un acento guardado en otro formato,
# con encoding_errors="replace" pandas ya lo puede leer
clientes = pd.read_csv("clientes.csv", encoding_errors="replace")

display(ventas)
display(clientes)
```

```python
# Inner: solo las ventas cuyo cliente sí está registrado
merge_inner = pd.merge(ventas, clientes, on="cliente_id", how="inner")

display(merge_inner)
```

```python
# Left: todas las ventas, aunque el cliente no esté registrado
merge_left = pd.merge(ventas, clientes, on="cliente_id", how="left")

display(merge_left)
```

```python
# Right: todos los clientes, aunque no hayan comprado nada
merge_right = pd.merge(ventas, clientes, on="cliente_id", how="right")

display(merge_right)
```

```python
# Outer: todas las ventas y todos los clientes, coincidan o no
merge_outer = pd.merge(ventas, clientes, on="cliente_id", how="outer")

display(merge_outer)
```

### Preguntas

**¿Qué tipo de merge mantuvo solo los clientes con compras?**

El inner. Solo deja las filas donde el `cliente_id` está en las dos tablas, o sea los clientes que sí tienen una venta. Por eso salen 4 filas.

**¿En qué merge aparecen ventas sin clientes registrados?**

En el left (y también en el outer). Como la tabla de la izquierda es `ventas`, se quedan todas las ventas. La venta 105 es del cliente 5, que no está en `clientes.csv`, y por eso sale con NaN en nombre, email y ciudad.

**¿Cuál merge permite ver clientes que no han comprado nada?**

El right (y también el outer). Ahí se quedan todos los clientes. Pedro Lara (cliente 6) sale con NaN en los datos de la venta porque no tiene ninguna compra.

**¿Cómo podrías usar merge para detectar clientes sin compras?**

Haciendo un merge tipo left con `clientes` como tabla de la izquierda y `ventas` a la derecha. Los clientes que no tienen ventas quedan con NaN en `venta_id`, así que solo hay que filtrar esas filas con `isnull()`.

```python
# Clientes que no tienen ninguna venta
clientes_ventas = pd.merge(clientes, ventas, on="cliente_id", how="left")
sin_compras = clientes_ventas[clientes_ventas["venta_id"].isnull()]

display(sin_compras)
```

**¿Cómo podrías identificar ventas sin un cliente registrado?**

Al revés: con el merge left que ya hicimos (`ventas` a la izquierda), las ventas cuyo cliente no existe quedan con NaN en `nombre`. Filtrando esas filas se ven las ventas sin cliente registrado.

```python
# Ventas cuyo cliente no está registrado
sin_cliente = merge_left[merge_left["nombre"].isnull()]

display(sin_cliente)
```
